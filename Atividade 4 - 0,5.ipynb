```python
import pandas as pd
import openpyxl
```

```python
arquivo_origem = r"C:\Users\enzo\OneDrive\Documentos\analise_dados\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
dados = pd.read_excel(arquivo_origem, engine="openpyxl", header=2)
```

```python
dados
```

```python
base_dados = dados.iloc[1:5570, 1:]
```

```python
base_dados
```

```python
base_dados.columns
```

```python
resumo_uf = (base_dados.groupby("UF")[
    [
        "População Município 2010\n(Sinopse)",
        "População 2010 (Alterações de Limites até 2022)1",
        "População Censo 2022"
    ]
    .sum()
    .reset_index())
```

```python
resumo_uf["variacao_populacional"] = (
    resumo_uf["População Censo 2022"]
    - resumo_uf["População Município 2010\n(Sinopse)"]
)
```

```python
resumo_uf = resumo_uf.sort_values(
    by="variacao_populacional", ascending=False
)
```

```python
resumo_uf.to_csv(
    r"C:\Users\enzo\OneDrive\Documentos\analise_dados\estados.csv",
    sep=";"
)
```
